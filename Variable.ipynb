```python
#UPPERCASE ALLOWED
NAME = "John"
print(NAME)
```

```python
#lowercase allowed
name = "John"
print(name)
```

```python
#combination of uppercase and lowercase allowed
Name = "John"
print(Name)
```

```python
#combination of chaters, numbers and underscores allowed
name1 = "John"
print(name1)
```

```python
#special characters not allowed
Name@ = "John"
print(Name@)
```

```python
#starting with numbers not allowed
1name = "John"  
print(1name)
```

```python
#only numbers not allowed
1 = "John"
print(1)
```

```python
#keywords not allowed
break = "John"
print(break)
```

```python
#spaces not allowed
my name = "John"
print(my name)
```
