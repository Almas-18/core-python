```python
x=10  # int
x
```

```python
#3:Tuple: its similar to the list but it is immutable. It means we cannot change the values of the tuple once it is created. Tuples are defined using parentheses ().
```

```python
t1=(1,2,3,4)
print(t1)
print(type(t1)) # Output: <class 'tuple'>   
```

```python
t1.remove(2) # This will raise an error because tuples are immutable and do not support item removal.
print(t1)
```

```python
#4:SET:its unique collection of data. It means it cannot have duplicate values. Sets are defined using curly braces {}.
```

```python
s1={1,2,3,4,5}
print(s1)
print(type(s1)) # Output: <class 'set'>
```

```python
s1={1,2,3,4,5}
s2={2,4,3,1,5}
print(s1==s2) # Output: True because the sets have the same elements
```

```python
s1.add(6) # Add 6 to the set
print(s1)
```
