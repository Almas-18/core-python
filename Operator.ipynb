# MEMBERSHIP OPERATOR:
                        1.IN
                        2.NOT IN

```python
x=[1,2,3,4,5]
print(3 in x) #check if 3 is present in the list
```

```python
x=[1,2,3,4,5]
print(6 not in x) #check if 6 is not present in the list
```

```python
str1="Hello"
print("H" in str1) #check if H is present in the string
```
