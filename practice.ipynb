```python
#2.Write a script that checks a student's score and assigns a grade: "A" for 90+, "B" for 80-89, "C" for 70-79, and "F" for anything below 70.
score=int(input("Enter the student score: "))
if score >90:
    print("A")
elif score>=80 and score<=89:
    print("B")
elif score>=70 and score <=79:
    print("C")
else:
    print("F")
```

```python
#3:3.Design a system that takes a person's age and prints "Child" (under 13), "Teenager" (13-19), or "Adult" (20+).
age=int(input("Enter the person's age:"))
if age<13:
    print("Child")
elif age >=13 and age<=19:
    print("Teenager")
else:
    print("Adult")
```

```python
#4.Write a program that checks the temperature in Celsius and prints "Freezing" (below 0), "Warm" (0-25), or "Hot" (above 25).
temp=int(input("Enter the temperature in Celsius:"))
if temp <0:
    print('freezing')
elif temp>=0 and temp<=25:
    print('warm')
else:
    print('hot')
```

```python
#Design a program that accepts a signal color ("Red", "Yellow", "Green") and prints "Stop", "Slow Down", or "Go".
signal_color=str(input("Enter the signal color:"))
if signal_color=='Red':
    print('stop')
elif signal_color=='Yellow':
    print('slow down')
elif signal_color=='Green':
    print('go')
else:
    print('invalid color')
```
