```python
l1=[20,100,300,50]
max=20
min=20
for i in l1:
    if i>max:
        max=i
    elif i<min:
        min=i
print("maximum value:",max)
print("minimum value:",min)
```

```python
start=1
end=9
for num in range(start,end+1):
    if num==1:
        continue
    prime=True
    for i in range(2,num):
      if num%i==0:
        prime=False
        break
    if prime:
        print(num)
```
