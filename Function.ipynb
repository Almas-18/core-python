```python
#WAP to check given number is even or odd
def Even(num):
    if num%2==0:
        return f"{num} is Even"
    else:
        return f"{num} is Odd"
Even(34)
```

```python
#WAP to check given number is greater than 50 or not
def greater(num):
    if num>50:
        return f"{num} is greater than 50"
    else:
        return f"{num} is not greater than 50"
greater(60)
```

```python
#WAP to check given multiple numbers for even and odd
def check_even_odd(numbers):
    result = {}
    for i in numbers:
        if i % 2 == 0:
            result[i] = "Even"
        else:
            result[i] = "Odd"
    return result
check_even_odd([10, 15, 22, 33, 40])
```
