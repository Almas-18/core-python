```python
# 7.	Check if a Number is a Palindrome
num=131
org=num
rev=0
while num>0:
    digit=num%10 
    rev=rev*10+digit
    num=num//10
if org==rev:
    print('palindrome')
else:
    print("not palindrome")
    
```
