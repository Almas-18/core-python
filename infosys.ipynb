```python
'''1. What is the question asking?
You are given vertical lines with different heights. Choose two lines that can hold the maximum amount of water.
The water capacity depends on two things:
- Width: distance between the two lines (right - left)
- Height: the shorter of the two lines
'''
def maxArea(heights):
    left = 0
    right = len(heights) - 1
    max_water = 0

    while left < right:
        width = right - left
        height = min(heights[left], heights[right])
        water = width * height

        max_water = max(max_water, water)

        if heights[left] < heights[right]:
            left += 1
        else:
            right -= 1

    return max_water


heights = [1, 6, 2, 5, 4, 8, 3, 7]
print(maxArea(heights))
    
    
```

```python
# Find the first character that appears again while reading the string from left to right.
def first_repeated(s):
    for i in range(len(s)):
        for j in range(i+1,len(s)):
            if s[i]==s[j]:
                return s[i] #Repeated character found; return it.
    return None #No repeated character found.
s='sanamm'
print(first_repeated(s))
```

```python
#longest substring without duplicate
def find(s):
    hashset=set()
    left=0
    maximum=0
    for right in range(len(s)):
        while s[right] in hashset:
            hashset.remove(s[right])
            left+=1
        hashset.add(s[right])
        maximum=max(maximum,right-left+1)
    return maximum
s='abcdabce'
print(find(s))
```
