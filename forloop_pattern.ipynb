```python
''' * #space 4 star=1 (" "*(num-i))+"*" *i=1*
   * * # space 3
  * * *
 * * * *
* * * * *
num=5

''' 
num=5

for i in range(1,num+1):
    print( " " * (num-i) +"* " * i)#" " * (num - i) → adds spaces before stars,increases stars: 1, 2, 3, 4, 5
```
