```python
#write a program to print 1 to 10 using while loop
i=1 #initialize the variable i with 1
while i <11: #while loop condition
    print(i,end=" ") #start
    i+=1 #increment i by 1 example :1+1=2, 2+1=3, 3+1=4, ..., 9+1=10, 10+1=11
```

```python
#WAP to reverse order of a number using while loop
i=10
while i>0:
    print(i, end=" ") #print the value of i in reverse order end " " is used to print the value of i in the same line
    i=i-1 #decrement i by 1 example :10-1=9, 9-1=8, 8-1=7, ..., 2-1=1, 1-1=0 
    
```

```python
#WAP To print table of 5
num=5
i=1
while i<=10: #i=1,i=2,i=3,...,i=10
    print(f"{num} * {i} = {num*i}") #print the table of 5 using f-string
    ''' 5*1=5
        5*2=10....5*10=50 '''
    i+=1 #increment i by 1 example :1+1=2, 2+1=3, 3+1=4, ..., 9+1=10, 10+1=11
```

```python
#2.Calculate the Sum of Digits of a Given Number
'''123 → 12 → 1 → 0
 ↓     ↓    ↓
 3     2    1

3 + 2 + 1 = 6'''
num=123
sum=0
while num>0:
    sum=sum+num%10 # 0+123%10=3  (123%3=3reminder)
    num=num//10 #123//10==12
print(sum)
```

```python
#7.Check if a Number is a Palindrom
num=131
org=num
rev=0
while num >0:
    digit=num%10
    rev=rev*10+digit
    num=num//10
if org==rev:
    print ("palin")
else:
    print("not")
```

```python
#8.Generate the Fibonacci Sequence Until a Specified Numbe
a=1
b=0
while a<=10:
    print(a)
    c=a+b
    b=a
    a=c
```

```python
#9.Calculate the Power of a Number Without Using the ** Operator
num=2
power=5
result=1
while power >0:
    result=result*num
    power=power-1
print(result)
```

```python
#11.Sum of Natural Numbers Until a Given Number
num=4
i=1
sum=0
while i<=num:
    sum=sum+i
    i=i+1
print(sum)
```

```python
#12.Implement a Simple Menu-Driven Program Using While Loop
while True:
    print("1.Add")
    print("2.sub")
    print("3.exit")
    choice=input("enter the choices:")
    if choice=="1":
        print("Add")
    elif choice=="2":
        print("sub")
    else:
        print("exit")
    break  
```
