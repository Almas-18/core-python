```python
'''1. String Frequency Analyzer
Question meaning

Find how many times each character appears.

Ignore spaces and treat uppercase/lowercase as the same.

Example:

Python Programming

P and p should be considered the same.'''
str1=input("Enter a string: ")

str1=str1.lower()
freq={}
for i in str1:
    if i == " ":
        continue
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
print(freq)
```

```python
'''2. Anagram Groups
Question meaning

Group words that contain the same characters.

Example:

["listen", "silent", "hello", "below", "elbow", "enlist"]

Output:

listen, silent, enlist
below, elbow
Easy approach'''

words = ["listen", "silent", "hello", "below", "elbow", "enlist"]

groups = {}  # Empty dictionary to store anagram groups

for word in words:
    # Sort the letters of the word
    key = "".join(sorted(word))

    # If this key is not present, create an empty list
    if key not in groups:
        groups[key] = []

    # Add the word to its group
    groups[key].append(word)

# Print only groups having more than one word
for group in groups.values():
    if len(group) > 1:
        print(group)
```

```python
#3. String Compression
# Question meaning
# If the same character appears continuously, replace it with:
# character + count

char = "aaabbcccccd"

result = ""       # Stores the final answer
count = 1         # Starts counting from 1

# Start from the second character
for i in range(1, len(char)):

    # Check if current and previous characters are same
    if char[i] == char[i - 1]:
        count += 1

    # If characters are different
    else:
        # Add previous character and its count
        result += char[i - 1] + str(count)

        # Start counting the new character
        count = 1

# Add the last character and its count
result += char[-1] + str(count)

print(result)
```

```python
'''4. Longest Word Without max()
Question meaning

Find the longest word in a sentence without using max().

If multiple words have the same longest length, print all of them.

Example:

Python is powerful and interesting

Output:

powerful, interesting'''

text="Python is powerful and interesting nnandiniiiiiiioi"
words = text.split()       # Split sentence into individual words

longest = []               # Store the longest word(s)
length = 0                 # Store the longest length

for word in words:         # Check each word
    if len(word) > length: # If word is longer than current longest
        length = len(word) # Update the longest length
        longest = [word]   # Store this word

    elif len(word) == length:  # If same longest length
        longest.append(word)   # Add this word too

print(longest)  

```

```python
'''6. Duplicate Frequency Analyzer
Question meaning

Find only the numbers that occur more than once and show their frequency.

Example:

[10, 20, 10, 30, 30, 20, 10, 40,30]

Output:

10 → 3
20 → 2
30 → 3'''
numbers = [10, 20, 10, 30, 20, 10, 40]

freq = {}  # Store number and its frequency

# Count each number
for i in numbers:
    if i in freq:
        freq[i] += 1  # Increase count if number already exists
    else:
        freq[i] = 1  # Add number with count 1

# Print numbers that appear more than once
for i in freq:
    if freq[i] > 1:
        print(f'{i} → {freq[i]}')
    
```

```python
'''7. Pair Sum Problem
Question meaning

Find pairs of numbers whose sum equals the target.

Example:

List = [2, 4, 3, 5, 7, 8, 9]
Target = 12

Pairs:

(4, 8)
(3, 9)
(5, 7)'''
numbers = [2, 4, 3, 5, 7, 8, 9]

target = 12

for i in numbers:                    # Take one number
    for j in numbers:                # Take another number
        if i + j == target and i < j: # Check sum and avoid duplicate pairs ex:8+4=12 and 4+8=12but here i=4 <j=8 so only 4+8=12 will be printed
            print(f"{i}, {j}")       # Print the pair
```

```python
'''8. Merge Three Lists
Question meaning

Combine three lists and create one list that:

Contains only unique elements
Is in ascending order
Does not use set()
Does not use sort() or sorted()

Example:

list1 = [10, 20, 30, 40]
list2 = [20, 40, 50, 60]
list3 = [30, 50, 70, 80]'''

list1 = [10, 20, 30, 40]
list2 = [20, 40, 50, 60]
list3 = [30, 50, 70, 80]

result = []                                      # Empty list

for i in list1 + list2 + list3:            # Take numbers from all 3 lists
    if i not in result:                       # Check if number is not already present
        result.append(i)                         # Add unique number

for i in range(len(result)):                    # i is used to repeat the sorting process.
    for j in range(len(result) - 1):             # Compare neighboring elements
        if result[j] > result[j + 1]:             # Check if current > next
            result[j], result[j + 1] = result[j + 1], result[j]  # Swap numbers

print(result)                                    # Print final sorted list
```

```python
'''11. Tuple Frequency Analysis
Question meaning

Find how many times each element occurs in a tuple.

Example:

numbers = (10, 20, 10, 30, 20, 10, 40)

Output:

10 → 3
20 → 2
30 → 1
40 → 1

The question says do not convert the tuple into a set.'''
numbers = (10, 20, 10, 30, 20, 10, 40)

freq = {}                                      # Empty dictionary

for i in numbers:                              # Take one number at a time
    if i in freq:                              # Check if number already exists
        freq[i] += 1                            # If yes, increase count by 1
    else:
        freq[i] = 1                             # If no, add number with count 1

for i in freq:                                 # Take each number from dictionary
    print(i, "->", freq[i])                    # Print number and its frequency

```

```python
'''12. Hollow Diamond Pattern
Question meaning

Print a diamond made using *, but the inside should be empty.

For n = 5:

    *
   * *
  *   *
 *     *
*       *
 *     *
  *   *
   * *
    *'''
num=5
for i in range(1,num+1):
    print(" "*(num-i),end=" ")
    if i==1:
        print("*")
    else:
        print("*"+" "*(2*i-3)+"*")
for i in range(num-1,0,-1):
    print(" "*(num-i),end=" ")
    if i ==1:
        print("*")
    else:
        print("*"+" "*(2*i-3)+"*")
```
